# **Competition Code**

## 1. Preprocessing Steps

````python
### Add the code related to the preprocessing steps in cells inside this section
import os
import re
import json
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score


file_paths = {}
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        full_path = os.path.join(dirname, filename)
        file_paths[filename] = full_path
print("找到的檔案路徑：")
for name, path in file_paths.items():
    print(f"{name}: {path}")

# 透過字典裡儲存的路徑來讀取資料 (DataFrame)
df_id = pd.read_csv(file_paths['data_identification.csv'])
df_emotion = pd.read_csv(file_paths['emotion.csv'])
df_posts = pd.read_json(file_paths['final_posts.json'])

print("\n資料讀取成功！")

# merge sets
df_merged = pd.merge(df_id, df_posts, on='post_id', how='left')
df_merged = pd.merge(df_merged, df_emotion, on='post_id', how='left')

# train/test sets
df_train = df_merged[df_merged['set'] == 'train'].reset_index(drop=True)
df_test = df_merged[df_merged['set'] == 'test'].reset_index(drop=True)
df_test = df_test.drop(columns=['label'])

print(df_train.shape)
print(df_test.shape)
print(df_train.head())
print("\n")
print(df_test.head())
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

# 設定圖表風格與尺寸
plt.figure(figsize=(10, 6))
sns.set_style("whitegrid")

# 計算各情緒類別的數量，並由大到小排序
label_counts = df_train['label'].value_counts()
total = len(df_train)

# 繪製長條圖
ax = sns.barplot(
    x=label_counts.index,
    y=label_counts.values,
    hue=label_counts.index,
    palette="viridis",
    legend=False
)

# 在每根長條上方加上數值與百分比標籤
for p in ax.patches:
    height = p.get_height()
    percentage = (height / total) * 100
    ax.annotate(
        f'{int(height)}\n({percentage:.1f}%)',
        (p.get_x() + p.get_width() / 2., height),
        ha='center',
        va='bottom',
        fontsize=10,
        xytext=(0, 5),
        textcoords='offset points'
    )

# 設定標題與座標軸標籤
plt.title('Emotion Class Distribution in Training Set', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Emotion Label', fontsize=12)
plt.ylabel('Number of Posts', fontsize=12)
plt.xticks(rotation=30)
plt.ylim(0, label_counts.max() * 1.15)  # 留出上方空間避免數字被切到

plt.tight_layout()
plt.show()
````

## 2. Feature Engineering Steps

````python
### Add the code related to the feature engineering steps in cells inside this section
def clean_text(text):
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    text = re.sub(r'\@\w+|\#', '', text) 
    text = re.sub(r'[^\w\s\U00010000-\U0010ffff]', '', text) #不清除掉表情符號
    return text

# clean the data and save in new column
df_train['clean_text'] = df_train['text'].apply(clean_text)
df_test['clean_text'] = df_test['text'].apply(clean_text)
````

````python
df_train['clean_text'] = df_train['text'].apply(clean_text)
mask = df_train['text'].str.contains(r'http|@|#', na=False)

for idx, row in df_train[mask].head(5).iterrows():
    print(f"=== Index: {idx} | Label: {row['label']} ===")
    print(f"[Raw]  : {row['text']}")
    print(f"[Clean]: {row['clean_text']}\n")
````

## 3. Model Implementation Steps

````python
### Add the code related to the model implementation steps in cells inside this section
# model construct
X_train, X_val, y_train, y_val = train_test_split(
    df_train['clean_text'], 
    df_train['label'], 
    test_size=0.15, 
    random_state=42, 
    stratify=df_train['label']
)

pipeline = Pipeline([
    # 使用 ngram_range=(1,2) 作為 Feature Engineering 的一部分
    ('tfidf', TfidfVectorizer(max_features=20000, 
                              ngram_range=(1, 5), 
                              min_df=2, 
                              sublinear_tf=True)),
    ('clf', LinearSVC(random_state=42, 
                      dual=False, 
                      max_iter=1000,
                      C=0.1,
                      class_weight='balanced'))
])

print("trianing... (TF-IDF + LinearSVC)")
pipeline.fit(X_train, y_train)

# 計算 Mean F1 Score (Macro F1)
y_pred_val = pipeline.predict(X_val)

val_mean_f1 = f1_score(y_val, y_pred_val, average='macro')
val_accuracy = accuracy_score(y_val, y_pred_val)

print(f"\nMean F1 Score: {val_mean_f1:.4f}")
print(f"Accuracy: {val_accuracy:.4f}")

pipeline.fit(df_train['clean_text'], df_train['label'])

# 預測測試集 (模型直接輸出 'anger', 'joy' 等字串，符合字串值要求)
test_predictions = pipeline.predict(df_test['clean_text'])

# 建立 DataFrame，並嚴格將欄位命名為 'id' 與 'label'
submission_df = pd.DataFrame({
    'id': df_test['post_id'], 
    'label': test_predictions
})

# 確保輸出無 index
submission_path = '/kaggle/working/submission.csv'
submission_df.to_csv(submission_path, index=False)
print(f"Save in {submission_path}")
````
